## 2. Structural Causal Model (SCM)

### Stage I: Specify the causal graph
A directed acyclic graph (DAG) `causal_graph` is explicitly defined using `networkx.DiGraph`. This graph represents the hypothesized causal relationships between variables such as 'OW', 'RI', 'OE', 'CC', 'SI', 'NI', and 'ET'.

### Stage II: Build and Fit the SCM
A `dowhy.gcm.StructuralCausalModel` is constructed based on the defined `causal_graph`. Causal mechanisms are assigned to each node:
*   Exogenous variables ('OW', 'SI', 'NI', 'OE') are assigned `EmpiricalDistribution`.
*   Endogenous variables ('RI', 'ET', 'CC') are assigned `AdditiveNoiseModel` with `linear_regressor` to model their dependence on their parents in the graph.

The model is then fitted to the `df` using `gcm.fit()`. The performance and consistency of the causal model are evaluated using `gcm.evaluate_causal_model()`, providing insights into KL divergence, MSE, R2 scores, and graph structure falsification.

---

## 3. Causal Graph Learning with GES

### GES Algorithm
The `causallearn` library's `ges` (Grow-Shrink) algorithm is applied to the dataset (`df`) with **default parameters** to learn a causal graph directly from data. The learned graph is then visualized and saved as `Openstack_GES.png`.

### GES Score Settings and Random Seed
*   The `ges` function was called with its default parameters. These typically include settings for the score function (e.g., BIC, BDeu) and convergence criteria.
*   The `numpy` random seed was set to `0` at the beginning of the notebook (`np.random.seed(0)`) to ensure reproducibility of operations involving random number generation, including the GES algorithm.

### Bootstrap Edge Stability
To assess the robustness of the learned causal edges, a bootstrap analysis is performed. The GES algorithm is run `n_bootstrap = 100` times on resampled datasets. The `edge_counts` matrix records how many times each edge appears across these bootstrap iterations, and `edge_stability` is calculated as the fraction of appearance. This stability matrix is then printed and visualized as a heatmap, providing confidence levels for the inferred causal links.

---

## 4. Model Interpretability with SHAP

### XGBoost Model Training
An `xgboost.XGBRegressor` model is trained to predict the target variable 'ET' using a subset of features `['CC', 'OE', 'OW', 'NI', 'SI', 'RI']` from `df`. The data is split into training (`X_train`, `y_train`) and testing (`X_test`, `y_test`) sets.

### Original SHAP Values
`shap.TreeExplainer` is used to calculate the SHAP (SHapley Additive exPlanations) values for the `X_test` data. These values quantify the contribution of each feature to the model's prediction for each instance. An initial `shap.summary_plot` is generated to visualize the global feature importance of the XGBoost model.

---

## 5. Causal SHAP Integration

### Causal Effect Estimation with DoWhy
Causal effects of 'RI' on 'ET' and 'CC' on 'ET' are estimated separately using `dowhy.CausalModel` with `backdoor.linear_regression` as the estimation method. The identified estimands and causal estimates are printed.

### Causal SHAP Values Adjustment
The original SHAP values for 'RI' and 'CC' are causally adjusted by multiplying them with their respective estimated causal effects obtained from DoWhy. This process aims to align the SHAP values with the true causal impact rather than just statistical association.

### Combined Causal SHAP Summary Plot
A combined causal SHAP summary plot is generated for all adjusted features, providing a visualization of feature importance considering the causal effects.

### Mean SHAP Values Calculation
Both the mean original SHAP values (`mean_shap_os`) and the mean causal SHAP values (`mean_causal_shap_values`) are calculated for comparison.

---

## 6. Refutation Tests

Refutation tests are performed for the causal effect of 'CC' on 'ET' using `dowhy` to check the robustness of the causal estimate. Specifically, 'random_common_cause' and 'data_subset_refuter' tests are applied.

---

## 7. Normalization and Global Importance

### Normalization
A helper function `normalize_shap_series` is defined to normalize SHAP values by dividing the absolute value of each SHAP value by the sum of absolute SHAP values for that series. This allows for comparison across different models or datasets. Both `mean_shap_os` and `mean_causal_shap_values` are normalized.

### Global Feature Importance
The normalized mean causal SHAP values from Openstack, along with similar values from Mozilla and Eclipse (if available and loaded into the notebook as `normalized_mean_shap_mz` and `normalized_mean_shap_ec`), are combined into a single DataFrame. A global mean importance is calculated across these datasets, and the results are printed and visualized as a bar chart, showing the overall importance of each feature.

````python
# Now fit the model
gcm.fit(causal_model, df)
````

````python
#bootstrap-based edge stability analysis
import numpy as np
import pandas as pd
from sklearn.utils import resample
from causallearn.search.ScoreBased.GES import ges

# Load your dataset (replace with actual file or DataFrame)
# Example: if you have a CSV file
#X = pd.read_csv("your_dataset.csv").values   # shape: (n_samples, n_features)

n_bootstrap = 100   # number of bootstrap iterations
#n_vars = X.shape[1]
n_vars = len(df.columns)
edge_counts = np.zeros((n_vars, n_vars))

for i in range(n_bootstrap):
    # Bootstrap resample (same number of rows, sampled with replacement)
    #X_resampled = resample(X, replace=True, n_samples=len(X))
    data_resampled = resample(df, replace=True, n_samples=len(df))
    X_resampled = data_resampled.to_numpy()

    # Run GES on resampled dataset
    ges_result = ges(X_resampled)
    adj_matrix = ges_result['G'].graph  # adjacency matrix (numpy array)

    # Count edges
    edge_counts += adj_matrix

# Compute stability scores (fraction of times each edge appears)
edge_stability = edge_counts / n_bootstrap

print("Edge stability matrix (values between 0 and 1):")
print(edge_stability)
````

````python
#Visualize Stability
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(edge_stability, annot=True, cmap="Blues", fmt=".2f")
plt.title("Bootstrap Edge Stability")
plt.show()
````

Similarly find the causal shap values for OW, Pr, SI, FI

````python
# Initialize causal_shap_values_all with the original SHAP values from XGBoost model
causal_shap_values_all = shap_values.copy()

# Get correct feature indices from X_test (XGBoost features)
ri_index = X_test.columns.get_loc('RI')
cc_index = X_test.columns.get_loc('CC')
ow_index = X_test.columns.get_loc('OW')
pr_index = X_test.columns.get_loc('Pr')
si_index = X_test.columns.get_loc('SI')
fi_index = X_test.columns.get_loc('FI')

# Apply causal adjustments cumulatively
if causal_estimate_os_RI.value is not None: # For RI
    causal_shap_values_all[:, ri_index] = shap_values[:, ri_index] * causal_estimate_os_RI.value
else:
    print("Warning: Causal effect for RI on ET could not be estimated. SHAP values for RI will not be causally adjusted.")

if causal_estimate_os_CC.value is not None: # For CC
    causal_shap_values_all[:, cc_index] = shap_values[:, cc_index] * causal_estimate_os_CC.value
else:
    print("Warning: Causal effect for CC on ET could not be estimated. SHAP values for CC will not be causally adjusted.")

if causal_estimate_os_OW.value is not None: # For OW
   causal_shap_values_all[:, ow_index] = shap_values[:, ow_index] * causal_estimate_os_OW.value
else:
   print("Warning: Causal effect for OW on ET could not be estimated. SHAP values for OW will not be causally adjusted.")

if causal_estimate_os_Pr.value is not None: # For Pr
   causal_shap_values_all[:, pr_index] = shap_values[:, pr_index] * causal_estimate_os_Pr.value
else:
  print("Warning: Causal effect for Pr on ET could not be estimated. SHAP values for Pr will not be causally adjusted.")

print("Causal SHAP values adjusted for RI, CC, OW, SI, FI and Pr.")
````

````python
# Calculate mean SHAP values
mean_shap_os = pd.Series(shap_values.mean(axis=0), index=X_test.columns)
mean_causal_shap_values = pd.Series(causal_shap_values_all.mean(axis=0), index=X_test.columns)

print("Mean Original SHAP Values:")
print(mean_shap_os)
print("\nMean Causal SHAP Values:")
print(mean_causal_shap_values)
````

````python
# Function to normalize SHAP values for a single dataset
def normalize_shap_series(series):
    abs_series = series.abs()
    # Handle cases where sum might be zero to avoid division by zero
    if abs_series.sum() == 0:
        return series * 0 # Return series of zeros if sum is zero
    return abs_series / abs_series.sum()

# Normalize mean SHAP values for OpenStack
normalized_mean_shap_os = normalize_shap_series(mean_shap_os)
# Similarfly calculate normalize mean SHAP values for Eclipse and Mozilla
normalized_mean_shap_mz = normalize_shap_series(mean_shap_mz)
normalized_mean_shap_ec = normalize_shap_series(mean_shap_ec)
````

````python
# Normalize mean causal SHAP values
normalized_mean_causal_shap_values = normalize_shap_series(mean_causal_shap_values)

print("Normalized Mean Original SHAP Values (Openstack):")
display(normalized_mean_shap_os)
print("\nNormalized Mean Causal SHAP Values (Openstack):")
display(normalized_mean_causal_shap_values)
````

````python
import matplotlib.pyplot as plt
# Combine the normalized mean causal SHAP values from Openstack, Mozilla, and Eclipse into a single DataFrame
all_models_mean_shap = pd.DataFrame({
    'Openstack Normalized Mean': normalized_mean_shap_os,
    'Mozilla Normalized Mean': normalized_mean_shap_mz,
    'Eclipse Normalized Mean': normalized_mean_shap_ec
})

# Calculate the overall mean importance for each feature across the available models
global_mean_importance = all_models_mean_shap.mean(axis=1).sort_values(ascending=False)

print("\n--- Global Feature Importance (Mean Across Openstack, Mozilla, and Eclipse - Normalized) ---")
display(global_mean_importance)

# Visualize the global feature importance
plt.figure(figsize=(10, 6))
global_mean_importance.plot(kind='barh', color='teal')
plt.title('Global Feature Importance (Mean Across Openstack, Mozilla, and Eclipse - Normalized)')
plt.xlabel('Mean Normalized Absolute SHAP Value')
plt.ylabel('Feature')
plt.gca().invert_yaxis() # Highest importance at the top
plt.tight_layout()
plt.show()
````
